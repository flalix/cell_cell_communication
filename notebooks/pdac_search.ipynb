```python
import requests, json, time
```

```python
URL_eutils="https://eutils.ncbi.nlm.nih.gov/entrez/eutils/"
```

```python
def esearch(db,term,n=500):
    r=requests.get(URL_eutils+"esearch.fcgi",params=dict(db=db,term=term,retmax=n,retmode="json"),timeout=30); r.raise_for_status()
    j=r.json()["esearchresult"]; return int(j["count"]), j["idlist"]
```

```python
qs={
 "pm_sc":'(pancreatic ductal adenocarcinoma[tiab] OR PDAC[tiab]) AND ("single-cell"[tiab] OR "single cell"[tiab] OR scRNA-seq[tiab] OR snRNA-seq[tiab] OR "single-nucleus"[tiab])',
 "pm_sp":'(pancreatic ductal adenocarcinoma[tiab] OR PDAC[tiab]) AND ("spatial transcriptom*"[tiab] OR Visium[tiab] OR "Xenium"[tiab] OR CosMx[tiab] OR MERFISH[tiab] OR "GeoMx"[tiab] OR "spatial multi-omic*"[tiab] OR "Slide-seq"[tiab] OR "Stereo-seq"[tiab])',
}

res={}
for k,q in qs.items():
    res[k]=esearch("pubmed",q,2000); print(k,res[k][0]); time.sleep(0.4)
```

```python
geoq={
 "geo_sc":'(pancreatic ductal adenocarcinoma OR PDAC OR pancreatic cancer) AND (single cell OR single-cell OR scRNA-seq OR snRNA-seq OR single nucleus) AND "Homo sapiens"[Organism] AND gse[ETYP]',
 "geo_sp":'(pancreatic ductal adenocarcinoma OR PDAC OR pancreatic cancer) AND (spatial OR Visium OR Xenium OR CosMx OR GeoMx OR MERFISH) AND gse[ETYP]',
 "geo_sc_mouse":'(pancreatic ductal adenocarcinoma OR PDAC OR KPC OR pancreatic cancer) AND (single cell OR single-cell OR scRNA-seq) AND "Mus musculus"[Organism] AND gse[ETYP]',
}

for k,q in geoq.items():
    res[k]=esearch("gds",q,2000); print(k,res[k][0]); time.sleep(0.4)
```

```python
for key, val in res.items():
    print(key, val)
```
