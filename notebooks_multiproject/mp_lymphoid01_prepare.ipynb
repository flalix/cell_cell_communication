#### Background (from ccc01)

This tutorial shows how to use **cell2location** method for spatially resolving fine-grained cell types by integrating 10X Visium data with scRNA-seq reference of cell types. Cell2location is a principled Bayesian model that estimates which combination of cell types in which cell abundance could have given the mRNA counts in the spatial data, while modelling technical effects (platform/technology effect, contaminating RNA, unexplained variance).

In the human lymph node, we resolve spatially interlaced cellular compartments, mapping rare B cell states and predicting putative cellular interactions relevant to the interferon response. These results present cell2location as a versatile tool for comprehensive mapping of tissue architecture.

#### Strategy

Comprehensive mapping of tissue cell architecture via integrated single cell and spatial transcriptomics.  

**The key principle is to first identify resident cell types based on single-cell RNA-sequencing (scRNA-seq) from dissociated tissues, to then map the identified cell types to their tissue positions in situ based on spatial transcriptomic profiles** [6].

Spatially resolved RNA-seq methods such as Visium / Spatial Transcriptomics [11], HDST [12] and Slide-seq [13], where mRNAs are positionally captured from thin tissue sections using microarrays or bead-arrays, enable high-throughput data acquisition and rely on simple histology and molecular biology protocols.

In the above workflow, the step of mapping cell types using **spatial transcriptomics poses several analytical challenges**. 

1. First, spatial RNA-seq measurements (i.e. locations) combine multiple cell types as array-based mRNA capture currently do not match cellular boundaries in tissues. Thus,
  - each spatial position corresponds to either several cell types (Visium, Tomo-Seq) [11, 14] or 
  - fractions of multiple cell types (Slide-Seq, HDST) [12, 13].
2. Second, spatial RNA-seq measurements are confounded by different sources of variation as 
  - cell numbers vary across tissue positions, 
  - different cells and cell types differ in total mRNA content [15], and
  - thin tissue sectioning captures variable fractions of each cell’s volume [16] 

Computational approaches need to appropriately model and account for all of these factors.

### Cell2location: a Bayesian model for spatial mapping of cell types

Cell2location maps the spatial distribution of cell types by integrating single-cell RNAseq (scRNA-seq) and multi-cell spatial 
transcriptomic data from a given tissue (Fig 1A). 
1. The first step of our model is to **estimate reference cell type signatures from scRNA-seq profiles**, for example as obtained **using conventional clustering to identify cell types and subpopulations followed by estimation of average cluster gene expression profiles** (Suppl. Methods, Fig S1). Cell2location implements this estimation step based on Negative Binomial regression, which allows to robustly combine data across technologies and batches (Suppl. Methods). 
2. In the second step, cell2location decomposes mRNA counts in spatial transcriptomic data using these reference signatures, thereby estimating the relative and absolute abundance of each cell type at each spatial location (Fig 1A, Fig S1).

**Cell2location is implemented as an interpretable hierarchical Bayesian model**, thereby
1. providing principled means to account for model uncertainty, 
2. accounting for linear dependencies in cell type abundances, 
3. modelling differences in measurement sensitivity across technologies, and 
4. accounting for unexplained/residual variation by employing a flexible count-based error model.
5. cell2location is computationally efficient, owing to variational approximate inference and GPU acceleration. 

For full details and a comparison to existing approaches see Suppl. Methods. The cell2location software comes with a suite of
downstream analysis tools, including the identification of groups of cell types with similar spatial locations.

### Cell2location

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/BayraktarLab/cell2location/blob/master/docs/notebooks/cell2location_tutorial.ipynb)

In this tutorial, we analyse a publicly available Visium dataset of the human lymph node from 10X Genomics, and spatially map a comprehensive atlas of 34 reference cell types derived by integration of scRNA-seq datasets from human secondary lymphoid organs.

- Cell2location provides high sensitivity and resolution by borrowing statistical strength across locations. This is achieved by modelling similarity of location patterns between cell types using a hierarchical factorisation of cell abundance into tissue zones as a prior (see paper methods).

- Using our statistical method based on Negative Binomial regression to robustly combine scRNA-seq reference data across technologies and batches results in improved spatial mapping accuracy. Given cell type annotation for each cell, the corresponding reference cell type signatures $g_{f,g}$, which represent the average mRNA count of each gene $g$ in each cell type $f$, can be estimated from sc/snRNA-seq data using either 1) NB regression or 2) a hard-coded computation of per-cluster average mRNA counts for individual genes.  We generally recommend using NB regression. This notebook uses a dataset composed of multiple batches and technologies. When the batch effects are small, a faster hard-coded method of computing per cluster averages provides similarly high accuracy. We also recommend the hard-coded method for non-UMI technologies such as Smart-Seq 2.

- Cell2location needs untransformed unnormalised spatial mRNA counts as input.

- You also need to provide cell2location with the expected average cell abundance per location which is used as a prior to guide estimation of absolute cell abundance. This value depends on the tissue and can be estimated by counting nuclei for a few locations in the paired histology image but can be approximate (see [paper methods for more guidance](https://github.com/BayraktarLab/cell2location/blob/master/docs/images/Note_on_selecting_hyperparameters.pdf)).

### 3GEX and 5GEX

3GEX and 5GEX refer to the two primary single-cell RNA sequencing (scRNA-seq) chemistries offered by 10x Genomics:
- 3' Gene Expression (3' GEX) and 
- 5' Gene Expression (5' GEX). 

Both assays measure whole-transcriptome gene expression at single-cell resolution, but they differ in which end of the mRNA transcript they capture. [1](https://kb.10xgenomics.com/s/article/360000939852-What-is-the-difference-between-Single-Cell-3-and-5-Gene-Expression-libraries), [2](https://ngisweden.scilifelab.se/methods/10x-genomics-3-or-5-gene-expression-gem-x-or-next-gem/)

#### Key Differences

| Feature | 3' Gene Expression (3GEX) | 5' Gene Expression (5GEX) |
| :--- | :--- | :--- |
| Captured Transcript End | Captures the 3' end (poly-A tail) of mRNA transcripts. | Captures the 5' end (5' UTR) of mRNA transcripts. |
| Gel Bead Oligo | Gel beads contain poly(dT) primers to bind the poly-A tail. | Gel beads contain a Template Switching Oligo (TSO), while poly(dT) is supplied in solution. |
| Primary Use Case | Standard whole-transcriptome profiling, cell atlasing, and general cell type discovery. | Immune profiling, specifically pairing gene expression with T-cell receptor (TCR) or B-cell receptor (BCR) V(D)J sequences. |
| Compatible Add-ons | Cell surface protein (CITE-seq), CRISPR screening, and CellPlex multiplexing. | Full-length V(D)J immune repertoire, BEAM antigen mapping, cell surface protein, and CRISPR screening. |
| Current Kit Architecture | GEM-X Universal 3' Gene Expression (v4). | GEM-X Universal 5' Gene Expression (v3). |

#### When to Choose Each Assay

- Choose 3GEX for general transcriptomics. If you do not need adaptive immune receptor sequences, 3' GEX is the standard workflow for characterizing cell populations, states, and differential gene expression across tissues. [1](https://www.10xgenomics.com/products/universal-three-prime-gene-expression), [2](https://kb.10xgenomics.com/s/article/360000939852-What-is-the-difference-between-Single-Cell-3-and-5-Gene-Expression-libraries)

- Choose 5GEX for immunology and V(D)J sequencing. Because the variable regions of TCRs and BCRs sit near the 5' end of the transcript, 5' GEX is required to capture full-length V(D)J sequences alongside gene expression from the exact same cell. [1](https://ngisweden.scilifelab.se/methods/10x-genomics-3-or-5-gene-expression-gem-x-or-next-gem/)
- Avoid mixing 3GEX and 5GEX in the same comparison without batch correction. Because the two chemistries capture different ends of transcripts and use different reverse transcription mechanics, combining them directly introduces technical batch effects that require computational integration (such as Harmony or Seurat). [1](https://www.sciencedirect.com/science/article/pii/S1525001626000213), [2](https://kb.10xgenomics.com/s/article/360000939852-What-is-the-difference-between-Single-Cell-3-and-5-Gene-Expression-libraries)
If you are planning an experiment or analyzing existing data, let me know:
- Your sample type (e.g., PBMCs, solid tissue, isolated nuclei)
- Whether you need immune receptor (TCR/BCR) data or multimodal readouts (surface protein/CRISPR)
I can help you narrow down the right kit configuration, sequencing depth, or Cell Ranger pipeline parameters.


### Batch correction?

The regression model fixed the problem only for the reference signatures. It did not fix it for the comparisons in the notebooks, which still mix 3′ and 5′ cells without any correction.

#### **What the model corrects.**

- fit_or_load_reference() runs RegressionModel.setup_anndata(batch_key="Sample", labels_key="Subset", categorical_covariate_keys=["Method"]). The loaded model confirms this: **23 batches**, 1 extra categorical covariate, and 2 chemistry levels (3GEX, 5GEX). **The model learns a separate 3′ vs 5′ effect for each gene, and the exported signatures have that effect removed. So the cell2location mapping on the Visium slide is corrected for chemistry.**

I also checked that the effect it learned is real. **Donor A16 is the only donor run on both chemistries**, so I compared 5′ with 3′ cells of the same tissue and subset there (42 groups). **The model's per-gene 5′/3′ log2 ratios correlate with those measured differences at Spearman 0.82**. The raw difference is large: 1,131 expressed genes differ more than 2-fold between chemistries in the same donor, tissue and cell type. Examples are *IGHM* (−2.35), *IGHD* (−2.19) and *JCHAIN* (−2.34).

#### **What it does not correct.**

All the descriptive comparisons in **lymphoid02** are computed from raw **X** normalised to log1p(CP10k). That covers the mLN age-group means (section 2), the tissue × age panels and dot plots (section 5), the AICDA by tissue/donor/chemistry analysis (section 6) and the subset markers. None of these functions in **single_cell_lymphoid_lib.py** include chemistry, and they never use the model's output.

#### **Where mixing biases the current results:**

- **Chemistry is tied to sample.** None of the 23 samples contains both chemistries, and only donor A16 does. Chemistry cannot be separated from donor except in A16.
- **mLN age groups:** "old adult" is a single donor (302c) and is 100% 3′. "Elder" is two 3′ donors (290b, 298c) plus two 5′ donors (390c, 417c). "Adult" (A16) is roughly half and half. Age differences in mLN are therefore partly chemistry differences.
- **Tonsil vs other tissues:** tonsil is 27,703 5′ vs 2,002 3′ cells, and has 12,907 of the 5′ GC/cycling B cells. The other tissues are close to 50/50 (A16). A comparison of tonsil against lymph node or spleen is mostly a 5′ against mixed-chemistry comparison. The AICDA conclusion ("not only in tonsils") also rests on about 60–120 GC B cells per tissue per chemistry outside the tonsil.

**Ways to fix it:**
1. **Compare within one chemistry.** For mLN, elder vs adult can be done within 3′ (290b, 298c vs A16-3′) and within 5′ (390c, 417c vs A16-5′). Old adult has no 5′ cells, so it can only be compared within 3′.
2. **Pseudobulk per sample** with Method in the design (e.g. **~ Method + age_group**). Sample is the unit, so chemistry nested within sample is not a problem. This is the cleanest way to test differences.
3. **Use a chemistry-aware model for expression values,** e.g. scVI with **batch_key="Sample"**, **categorical_covariate_keys=["Method"]** and **transform_batch**, if you want corrected plots or UMAP panels.

Don't use the model's per-gene factors to correct log1p(CP10k) values by hand. For several panel genes the model and the A16 data disagree in sign (e.g. *MS4A1* +0.67 vs −0.82, *CD19* +0.64 vs −1.25), partly because the model's factor is only defined up to the other scaling terms. Both values for each gene are in the table.

<div class="alert alert-info">
<b> Important </b>
    
To use cell2location spatial mapping model, you need to specify 2 user-provided hyperparameters (**N_cells_per_location** and **detection_alpha**) 

For detailed guidance on setting these hyperparameters and their impact see [the flow diagram and the note](https://github.com/BayraktarLab/cell2location/blob/master/docs/images/Note_on_selecting_hyperparameters.pdf).
</div>

**Choosing hyperparameter **N_cells_per_location****

It is useful to adapt the expected cell abundance **N_cells_per_location** to every tissue. This value can be estimated from paired histology images and as described in the note above. Change the value presented in this tutorial (**N_cells_per_location=30**, params['c2l_N_cells_per_location']) to the value observed in your tissue.

**Choosing hyperparameter **detection_alpha****

To improve accuracy & sensitivity on datasets with large technical variability in RNA detection sensitivity within the slide/batch - you need to relax regularisation of per-location normalisation (use **detection_alpha=20**). High technical variability in RNA detection sensitivity is present in your sample when you observe the spatial distribution of total RNA count per location that doesn't match expected cell numbers based on histological examination.

We initially opted for high regularisation (**detection_alpha=200**) as a default because the mouse brain & human lymph node datasets used in our paper have low technical effects and using high regularisation strength improves consistency between total estimated cell abundance per location and the nuclei count quantified from histology ([Fig S8F in cell2location paper](https://static-content.springer.com/esm/art%3A10.1038%2Fs41587-021-01139-4/MediaObjects/41587_2021_1139_MOESM1_ESM.pdf)). However, in many collaborations, we see that Visium experiments on human tissues suffer from technical effects. This motivates the new default value of **detection_alpha=20** and the recommendation of testing both settings on your data (**detection_alpha=20** and **detection_alpha=200**).

### Naive CD4⁺ and CD8⁺ T cells share the paracortex in a lymph node, so they should map to the same spots.

**Why T_CD4+_naive and T_CD8+_naive are in the same region.** Mostly real biology, made stronger by the method:

| check on the mapped slide | value |
|---|---|
| correlation of the two abundances across the 4,035 spots (Spearman ρ) | 0.98 |
| share of total abundance in the T-zone regions (5, 9, 12, 13, 16, 18; 22% of spots) | 62% for CD4, 61% for CD8 |
| CD4 : CD8 naive ratio, T zone / rest of the slide | 4.1 / 3.9 |
| weight on the same NMF factor (fact_1) | 0.92 for CD4, 0.81 for CD8 |

1. **Biology.** Naive CD4⁺ and CD8⁺ T cells both use CCR7 and CD62L (*SELL*) to enter the node and settle in the paracortex. They mix there, with CD4 outnumbering CD8 by roughly 2–4 to 1. Nothing in the tissue gives naive CD8 cells a separate zone at this scale, so they really are in the same region. The other cell types back this up: both correlate with T_Treg (ρ ≈ 0.74) and both are excluded from follicles (ρ ≈ −0.5 with B_GC_LZ).

2. **Resolution.** A Visium spot is 55 µm across and holds roughly 10–30 cells. Any T-zone spot therefore contains both kinds of naive T cell, so even a small-scale CD4/CD8 separation, if it exists, would be averaged out.
3. **The method makes the maps look even more alike:**

   - The two reference signatures differ in only a handful of genes, mainly *CD4*, *CD8A* and *CD8B*, and these are weakly detected. *CD8A* correlates only slightly more with the CD8 map (ρ 0.29 vs 0.22), and *CD4* is the same for both (0.21).
   
   - Cell2location's prior groups cell types that occur together into shared tissue zones, which pulls their maps closer.
   - As a result, the CD8 share of naive T cells in T-zone spots barely varies: median 0.20, middle half 0.18–0.21.

**There is still a CD8-specific signal.** Within T-zone spots, the CD8 share rises with *CD8A* + *CD8B* expression (ρ = 0.46) and not with *CD4* (ρ = −0.15). So the model does split the two by marker genes, but only as small changes in proportion inside one shared zone. If you want maps that look different, plot the CD8 share of naive T cells rather than the two abundances.

### 8. Downstream analysis: regions

#### Identifying discrete tissue regions by Leiden clustering <a class="anchor" id="Identifying-discrete-tissue-regions-by-Leiden-clustering"></a>

We identify tissue regions that differ in their cell composition by clustering locations using cell abundance estimated by cell2location.

We find tissue regions by clustering Visium spots using estimated cell abundance each cell type. We construct a K-nearest neighbour (KNN) graph representing similarity of locations in estimated cell abundance and then apply Leiden clustering. The number of KNN neighbours should be adapted to size of dataset and the size of anatomically defined regions (i.e. hippocampus regions are rather small compared to size of the brain so could be masked by large **n_neighbors**). This can be done for a range KNN neighbours and Leiden clustering resolutions until a clustering matching the anatomical structure of the tissue is obtained.

The clustering is done jointly across all Visium sections / batches, hence the region identities are directly comparable. When there are strong technical effects between multiple batches (not the case here) **sc.external.pp.bbknn** can be in principle used to account for those effects during the KNN construction.

The resulting clusters are saved in **adata_vis.obs['region_cluster']**.

**scl.cluster_regions**: kNN (15 neighbours) on the q05 abundances, Leiden resolution 1.1 with the leidenalg backend (scanpy default) and seed 0 (**params['region_leiden_flavor']**; 'igraph' is faster but gives different clusters). **scl.umap_regions**: UMAP on that graph (min_dist 0.3, spread 1).

### 10. NMF co-location

- Here, we use the cell2location mapping results

  - to identify the spatial co-occurrence of cell types 
  - in order to better understand the tissue organisation and 
  - predict cellular interactions. 

- We performed non-negative matrix factorization (NMF) of the cell type abundance estimates from cell2location ([paper section 4, Fig 4D](https://www.nature.com/articles/s41587-021-01139-4)).

- Similar to the established benefits of applying NMF to conventional scRNA-seq, the additive NMF decomposition yielded a grouping of spatial cell type abundance profiles into components that capture co-localised cell types ([Supplementary Methods section 4.2, p. 60](https://www.nature.com/articles/s41587-021-01139-4#Sec50)). 

- This NMF-based decomposition naturally accounts for the fact that multiple cell types and microenvironments can co-exist at the same Visium locations (see [paper Fig S20, p. 34](https://www.nature.com/articles/s41587-021-01139-4#Sec50)), while sharing information across tissue areas (e.g. individual germinal centres).


<div class="alert alert-block alert-primary">
<b>Tip</b>
    
In practice, it is better to train NMF for a range of factors $R={5, .., 30}$ and select $R$ as a balance between capturing fine-grained and splitting known well-established tissue zones.

If you want to find a few most distinct cellular compartments, use a small number of factors.   

If you want to find very strong co-location signal and assume that most cell types don't co-locate, use a lot of factors (> 30). Here n_fact = 11 and 12 are used (see **n_facts** below).
</div>

### What is the relationship of CD3D and T-CD4+ naive?


![CD4_CD3E_CD3D](../figures/CD4_CD3E_CD3D.gif)

Ref: https://www.creative-biolabs.com/adc/target-cd3d-118.htm

CD3D is a general marker for all T cells. It is not specific to naive CD4⁺ T cells, but those cells cannot develop or work without it.

#### What CD3D is
- CD3D encodes **CD3δ**, one of the invariant chains of the **TCR–CD3 complex** (CD3γ, δ, ε and ζ alongside the TCRαβ heterodimer).
- CD3δ is needed for the TCR complex to assemble and reach the cell surface.
- Its ITAM motifs pass the signal on after antigen recognition, through the LCK → ZAP70 → LAT/SLP-76 pathway.

#### How it relates to naive CD4⁺ T cells
1. **Development depends on it.** 
- CD3δ is needed for pre-TCR and TCR signalling in the thymus, especially for positive selection of CD4⁺CD8⁺ thymocytes. 
- People with CD3D deficiency have T⁻B⁺NK⁺ severe combined immunodeficiency (SCID): 
  - they have almost no mature αβ T cells, so no naive CD4⁺ T cells either.
2. **Activation depends on it.** 
- A naive CD4⁺ T cell is activated for the first time when its TCR recognises a peptide on MHC-II. 
- That signal passes through the CD3 chains, so CD3δ is part of the machinery that turns naive cells into effector or memory cells.
3. **It is a lineage marker, not a state marker.** 
- In single-cell RNA-seq, CD3D (with CD3E and CD3G) is one of the most reliable transcripts for marking a cluster as T cells. 
- It is expressed in:
   - naive and memory CD4⁺ T cells
   - CD8⁺ T cells
   - γδ T cells and MAIT cells
   - some NKT/ILC-like cells

   It is typically **high in naive T cells**. However, it cannot tell naive cells from memory cells, or CD4⁺ cells from CD8⁺ cells.

#### Defining a "T-CD4⁺ naive" cluster
Annotation is usually layered:

| Level | Markers |
|---|---|
| T-cell identity | **CD3D**, CD3E, CD3G, TRAC |
| CD4 lineage | CD4, CD40LG (absence of CD8A/CD8B) |
| Naive state | CCR7, SELL (CD62L), LEF1, TCF7, IL7R, MAL, PTPRC-RA isoform (CD45RA protein) |
| Should be low | CD44, GZMK/GZMB, CCL5, IFNG, PRF1, HLA-DR, FAS/CD95, PTPRC-RO isoform (CD45RO protein) |

So in a dot plot or marker table, CD3D should be high in the T-CD4⁺ naive cluster. It should be just as high in the other T-cell clusters, so it confirms "T cell", not "naive CD4". The naive label rests on CCR7, SELL, LEF1 and TCF7 with low effector genes.

### What is the relationship of CR2 and B-CELL germinal center light zone?

CR2 (CD21) is a light-zone gene, but most of it comes from follicular dendritic cells (FDCs), the stromal cells that form the light-zone network, rather than from the germinal-centre B cells. On B cells it is a co-receptor that helps them win the antigen-driven selection that happens in the light zone.


![bcell_receptor_and_CR2.jpg](../figures/bcell_receptor_and_CR2.jpg)

Ref: https://www.sciencedirect.com/topics/nursing-and-health-professions/complement-receptor

#### What CR2 is
- CR2 encodes **CD21 (complement receptor 2)**, which binds the complement fragments C3d, C3dg and iC3b that coat antigens and immune complexes.
- On B cells it forms the **CD21–CD19–CD81 co-receptor complex**. When it binds C3d on an antigen at the same time as the BCR binds that antigen, the BCR signal gets much stronger: C3d-tagged antigen can be up to about 1000-fold more immunogenic.
- It is also the entry receptor for **Epstein–Barr virus (EBV)**.

#### The connection to the light zone
1. **FDCs are the main CR2 source in the light zone.** FDCs express very high CD21, mainly the long isoform, together with CD35 (CR1) and Fc receptors. They use these receptors to hold complement-coated immune complexes on their surface for long periods. In histology and spatial transcriptomics, a CR2/CD21 signal is a standard way to find the light zone, alongside FDCSP, CXCL13, CLU and FCER2.
2. **Antigen is presented to light-zone B cells.** After mutating their antibody genes in the dark zone, B cells move to the light zone and test their BCRs on antigen held by FDCs. B cells with higher affinity take up more antigen, show more peptide–MHC-II to T follicular helper (Tfh) cells, and get CD40L/IL-21 help. That help drives them back to the dark zone or into plasma-cell or memory fates.
3. **CD21 on B cells helps with selection.** Because the antigen on FDCs is complement-coated, CD21 on the B cell binds it alongside the BCR, which lowers the activation threshold. Mice lacking Cr2 have smaller, shorter-lived germinal centres, less affinity maturation and fewer memory B cells. Experiments that swap which cells carry Cr2 show that both FDC and B-cell CD21 contribute.
4. **Germinal-centre B cells themselves have relatively low CD21.** GC B cells express less CD21 than naive or follicular mantle B cells. Within the germinal centre, light-zone cells tend to have somewhat more than dark-zone cells, but CR2 is not a reliable marker for separating the two zones.

Immune Complex (IC) acquisition. In the spleen (left panel), non-cognate marginal zone (MZ) B cells capture ICs, consisting of antigen, IgM or IgG, and C3 degradation products (C3d and C3b), from the blood stream (1). MZ B cells, which have captured ICs in a C3d/CR2-dependent fashion, then migrate into the follicle, where they transfer the ICs onto FDCs, which bind them via C3b and CR1 (2). The ICs are then released from the MZ B cell (3). In the lymph node (right panel), subcapsulary sinus macrophages (SCSMϕ) capture ICs consisting of antigen, IgM or IgG, and C3d and C3b degradation products from the lymph (1). SCSMϕ migrate into the follicle and transfer ICs to follicular (FO) B cells in a CR2-dependent manner (2). Subsequently, FO B cells transfer the ICs onto FDCs (3, 4).

Ref: Kranich J, Krautler NJ. How Follicular Dendritic Cells Shape the B-Cell Antigenome. Front Immunol. 2016 Jun 21;7:225. doi: 10.3389/fimmu.2016.00225. PMID: 27446069; PMCID: PMC4914831.



#### Practical implications for annotation

| Context | What CR2 tells you |
|---|---|
| Spatial / histology (Visium, IHC, IF) | Strong marker of the **light-zone FDC network** |
| scRNA-seq, stromal cells | High CR2 together with FDCSP, CXCL13 and CLU means **FDCs** |
| scRNA-seq, B cells | Higher in naive/mantle and memory B cells than in GC B cells; not a defining marker for light-zone B cells |
| Better light-zone vs dark-zone B-cell markers | **Light zone:** CD83, CD86, BCL2A1, NFKBIA, MYC (subset), CIITA/HLA-II, CD40 target genes. **Dark zone:** CXCR4, AICDA, MKI67, AURKA, FOXP1 |
