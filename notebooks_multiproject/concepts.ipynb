### Compressed Sparse Row (CSR)

A Compressed Sparse Row (CSR) matrix is a storage format that saves memory and speeds up calculations by storing only the non-zero elements of a sparse matrix. [1](https://medium.com/data-science/handling-sparse-matrix-concept-behind-compressed-sparse-row-csr-matrix-4fe6abe58a7a)

#### The Three Arrays of CSR

CSR represents an M \times \nasty or M × N matrix using three 1-D arrays: [1](https://en.wikipedia.org/wiki/Sparse_matrix)

1. data (values): A 1-D array of length NNZ (Number of Non-Zeros) containing all explicit non-zero values, listed row by row.

2. indices (column_indices): A 1-D array of length NNZ containing the column index for each corresponding value in the data array.
3. indptr (row_pointers / row offsets): A 1-D array of length M + 1. It points to the start index in data and indices where each row begins. The last element equals NNZ. For row i, the non-zero elements are located from index indptr[i] to indptr[i+1] - 1. [1](https://docs.scipy.org/doc/scipy/reference/generated/scipy.sparse.csr_matrix.html), [2](https://bluuewhale.github.io/posts/sparse-matrix-and-csr-en/), [3](https://www.youtube.com/watch?v=H6YGKNukGMo&t=2188), [4](https://docs.nvidia.com/nvpl/latest/sparse/storage_format/sparse_matrix.html), [5](https://www.youtube.com/watch?v=hlTwhuamoS0)

#### Key Advantages & Limitations

• Pros:
	• Memory efficient: Avoids storing explicit zero values or repeating row numbers.
	• Fast operations: Excels at row-based access, slicing, and matrix-vector multiplications (Mx). [1](https://www.youtube.com/watch?v=H6YGKNukGMo&t=2188), [2](https://bluuewhale.github.io/posts/sparse-matrix-and-csr-en/), [3](https://en.wikipedia.org/wiki/Sparse_matrix)

• Cons:
	• Slow column slicing/access: Finding all non-zeros in a specific column is inefficient.
	• Expensive modifications: Inserting or removing non-zero elements requires shifting arrays and updating row pointers. [1](https://scipy-lectures.org/advanced/scipy_sparse/csr_matrix.html), [2](https://www.youtube.com/watch?v=H6YGKNukGMo&t=2188), [3](https://www.youtube.com/watch?v=hlTwhuamoS0)

## Cell2location

### **Reference cell types** 

- Reference cell types are the cell types annotated in the **single-cell reference**:

  - the scRNA-seq dataset whose cells were sequenced one by one and labelled by the authors. 

  - In this pipeline they're the names that **cell2location (or RCTD / stereoscope)** maps onto the Visium spots.

#### **Why a reference is needed:** 

A Visium spot (55 µm) holds roughly 10–30 cells of several types, and its RNA is a mixture. 

Deconvolution works in two steps:

1. **Signatures:** from the reference, learn the expression profile of each annotated type (**fit_or_load_reference**, the q50 signatures in **sc.h5ad**).

2. **Mapping:** estimate how many cells of each of those types sit in every spot (**fit_or_load_mapping**, the q05 abundances).


#### So the "reference cell types" are the columns of every abundance table (**df_abund**)

The rows of the NMF dot plots, and the types the regions are named after.

**In each project:**

| project | single-cell reference | reference cell types |
|---|---|---|
| heart_dvlp | **hdca_heart_sc_reference.h5ad**, 76,991 cells | 31 high-level types (Mat_vCM, Immat_CM, Valve_MC, EPDC, …); the 4 **HL_excl_*** clusters were excluded by the authors |
| pdac | sc/sn reference | 42 types (D01_Ductal-CFTR, F01_myCAF-MMP11, …), grouped into 4 lineages |
| lymphoid | **lymphoid_organ_sc.h5ad** | 34 types (B_GC_LZ, T_CD4+_TfH, FDC, …) |
| mouse_brain | snRNA reference | 59 subtypes |

#### **Consequence** 

The spots can only be described with these types. A cell type missing from the reference can't be found in the tissue. That's why the heart's haemoglobin-rich "blood" regions had no matching reference type.

### **spot deconvolution** methods. 

- Given a single-cell reference with annotated cell types, each estimates how much of each type is in every Visium spot. 
- They differ in the statistical model, in what they output (proportions or cell counts), and in whether they use the spots' spatial neighbourhood.

**Reference-based deconvolution (same job as cell2location):**

| method | model | output | language / package | notes |
|---|---|---|---|---|
| **cell2location** | Bayesian negative binomial; reference signatures by NB regression; per-section and per-gene technical effects | **absolute abundance** (cells per spot) | Python (scvi-tools) | needs a GPU for large data; used here for heart, lymphoid, mouse_brain |
| **RCTD** | Poisson / log-normal likelihood with platform-effect correction; **full**, **doublet** or **multi** mode | **proportions** (weights, sum to 1) | R (**spacexr**) | fast on CPU; used here for pdac |
| **stereoscope** | negative binomial; reference parameters per type, then a mixture per spot | proportions | Python (originally standalone, now in scvi-tools) | what the heart paper (Lázár et al.) used |
| **DestVI** | VAE (variational autoencoder) on both data sets; also gives a continuous cell *state* within each type | proportions + state | Python (scvi-tools) | for gradients within a type |
| **CARD** | NMF regression with a spatial CAR (conditional autoregressive) prior, so neighbouring spots share information | proportions | R | can also impute a higher-resolution map |
| **SPOTlight** | NMF topic signatures + non-negative least squares | proportions | R | fast and simple |
| **SpatialDWLS** | marker-gene enrichment, then dampened weighted least squares | proportions | R (Giotto) | |
| **Seurat anchors** | label transfer (anchor-based integration) | prediction scores | R | not a true mixture model |
| **Tangram** | maps single cells onto spots by optimisation | cell → spot assignment | Python | also projects genes into space |
| **CytoSPACE** | assigns reference cells to spots, using a cells-per-spot estimate | cell assignment | Python | gives single-cell-resolution maps |

**Without a reference:**

- **STdeconvolve** (R) finds cell-type-like "topics" in the spots with LDA (latent Dirichlet allocation). You name them afterwards from their marker genes.


**Benchmarks:** 

in independent comparisons (e.g. Li et al., Nat Methods 2022; the Spotless benchmark, Sang-aram et al., eLife 2024), **cell2location and RCTD** are consistently among the most accurate. SpatialDWLS, stereoscope and CARD also do well. Results depend on the tissue, the sequencing depth and the quality of the reference.

**Practical differences that matter in your projects:**
- **Cell counts vs shares:** 

  - cell2location gives cell counts per spot, while 
  
  - RCTD and stereoscope give shares that sum to 1. 
  - That's why the RCTD weights had to be rescaled before the NMF co-location, and why regions clustered on c2l counts also split by sequencing depth (the heart "depth tiers"), which share-based regions don't.

- **Batch effects:** cell2location models technical differences between sections explicitly, which suits multi-section studies like the 38 heart sections.
- **Compute:** RCTD runs on a CPU in minutes; cell2location needed the GPU and about 5 h for the heart.

### benchmarks

Yes, several independent benchmarks exist. None names a single winner for every case, but they agree on a short list, and **cell2location and RCTD** are at or near the top in essentially all of them.

**Main benchmarks:**
- **Spotless (Sang-aram et al., eLife 2024):** compared eight spatial deconvolution methods (cell2location, DestVI, DSTG, RCTD, SpatialDWLS, SPOTlight, stereoscope, STRIDE), plus a bulk method and two mapping methods. They found that cell2location and RCTD are the top-performing methods, but surprisingly, a simple regression model outperforms almost half of the dedicated spatial deconvolution methods. Methods with better performance – cell2location, RCTD, SpatialDWLS, and stereoscope – were also more stable against changing reference datasets.

- **Li et al., Bioinformatics 2023 (14 methods):** cell2location, RCTD and spatialDWLS are more accurate than other ST deconvolution methods; cell2location and spatialDWLS are more robust to the variation of sequencing depth than RCTD; the accuracy of the existing methods tends to decrease as the spot size becomes smaller.
- **Li et al., Nature Communications 2023 (18 methods, 50 datasets):** In general, CARD, Cell2location, Tangram, and RCTD were the best performing methods.
- **A 2026 preprint across tissues and technologies:** it identified Cell2location, RCTD, and SONAR as the top-performing methods. Across 13 cancer datasets, Cell2location, RCTD, and Stereoscope emerged as the top-performing methods. On speed: among the four most accurate methods, Cell2location, RCTD, SONAR, and CARD, Cell2location exhibited the longest runtimes, whereas RCTD and CARD were the fastest.
- **A review summarising the benchmarks:** the consensus thus far is that cell2location, RCTD and stereoscope are the generally superior methods, followed by spatial DWLS. The rankings differ between studies due to a combination of reasons, including different reference datasets, testing datasets, gold standards as well as evaluation metrics.

**So which is best?** There's no single best. The practical rankings:
1. **cell2location and RCTD:** consistently top-tier. cell2location is slower (needs a GPU) but gives absolute cell counts and handles differences in sequencing depth well. RCTD is fast on a CPU and gives proportions.

2. **stereoscope, SpatialDWLS, CARD:** usually close behind. CARD is fast and uses the spatial neighbourhood.
3. **Many others:** often no better than a simple regression baseline.

Performance also varied substantially based on tissue architecture, spatial technology, dataset scale, and cell type diversity. The quality of your single-cell reference matters as much as the choice of method.

**Your projects** use benchmark-leading methods: cell2location for heart, lymphoid and mouse_brain; RCTD for pdac; and the heart paper's stereoscope as the comparison. One sensible check, which is planned for heart (`calc_df_c2l_vs_stereoscope`), is to compare two top methods on the same data: cell types where they agree are robust, and those where they disagree need care.
