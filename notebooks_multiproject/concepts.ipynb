### Compressed Sparse Row (CSR)

A Compressed Sparse Row (CSR) matrix is a storage format that saves memory and speeds up calculations by storing only the non-zero elements of a sparse matrix. [1](https://medium.com/data-science/handling-sparse-matrix-concept-behind-compressed-sparse-row-csr-matrix-4fe6abe58a7a)

#### The Three Arrays of CSR

CSR represents an M \times \nasty or M × N matrix using three 1-D arrays: [1](https://en.wikipedia.org/wiki/Sparse_matrix)

1. data (values): A 1-D array of length NNZ (Number of Non-Zeros) containing all explicit non-zero values, listed row by row.

2. indices (column_indices): A 1-D array of length NNZ containing the column index for each corresponding value in the data array.
3. indptr (row_pointers / row offsets): A 1-D array of length M + 1. It points to the start index in data and indices where each row begins. The last element equals NNZ. For row i, the non-zero elements are located from index indptr[i] to indptr[i+1] - 1. [1](https://docs.scipy.org/doc/scipy/reference/generated/scipy.sparse.csr_matrix.html), [2](https://bluuewhale.github.io/posts/sparse-matrix-and-csr-en/), [3](https://www.youtube.com/watch?v=H6YGKNukGMo&t=2188), [4](https://docs.nvidia.com/nvpl/latest/sparse/storage_format/sparse_matrix.html), [5](https://www.youtube.com/watch?v=hlTwhuamoS0)

#### Key Advantages & Limitations

• Pros:
	• Memory efficient: Avoids storing explicit zero values or repeating row numbers.
	• Fast operations: Excels at row-based access, slicing, and matrix-vector multiplications (Mx). [1](https://www.youtube.com/watch?v=H6YGKNukGMo&t=2188), [2](https://bluuewhale.github.io/posts/sparse-matrix-and-csr-en/), [3](https://en.wikipedia.org/wiki/Sparse_matrix)

• Cons:
	• Slow column slicing/access: Finding all non-zeros in a specific column is inefficient.
	• Expensive modifications: Inserting or removing non-zero elements requires shifting arrays and updating row pointers. [1](https://scipy-lectures.org/advanced/scipy_sparse/csr_matrix.html), [2](https://www.youtube.com/watch?v=H6YGKNukGMo&t=2188), [3](https://www.youtube.com/watch?v=hlTwhuamoS0)

```python
import numpy as np
from scipy.sparse import csr_matrix

# Construct using (data, (row_ind, col_ind))
data = np.array([1, 7, 5, 3])
row_ind = np.array([0, 0, 1, 2])
col_ind = np.array([0, 1, 0, 2])

matrix = csr_matrix((data, (row_ind, col_ind)), shape=(4,4))

print(matrix.toarray())
```
